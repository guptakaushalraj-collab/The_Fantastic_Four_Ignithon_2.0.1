## 6. Run your own evidence (optional)

Run the cell, choose your files (screenshots, PDFs, CSV exports, chat exports, SMS text), and the report is printed below it.

```python
import os, shutil
from google.colab import files

shutil.rmtree('my-case', ignore_errors=True)
os.makedirs('my-case')
for name, data in files.upload().items():
    with open(os.path.join('my-case', name), 'wb') as f:
        f.write(data)

!python -m fraud_evidence my-case --case-id MY-CASE
print(open('reports/MY-CASE/report.txt', encoding='utf-8').read())
```
